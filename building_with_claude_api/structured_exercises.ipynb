````python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-5"
````

````python
# temperatura en 1 es mucho mas creativa...

def chat(messages, stop_sequences=None, temperature=0.0):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature
    }
    
    if stop_sequences:
        params["stop_sequences"] = stop_sequences
        
    response = client.messages.create(**params)
    return response.content[0].text
````

````python
def add_user_message(messages, text):
    user_message = { "role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(message, text):
    assistant_message = { "role": "assistant", "content": text}
    messages.append(assistant_message)
````

````python
# ejercicio
# 
messages = []

add_user_message(messages, "Generate three different sample AWS CLI commands. Each should be very short")

answer = chat(messages)

answer

# Ojo no queremos esos caracteres adicionales, solo queremos el json...
````
