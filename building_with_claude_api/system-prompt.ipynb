```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
def add_user_message(messages, text):
    user_message = { "role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(message, text):
    assistant_message = { "role": "assistant", "content": text}
    messages.append(assistant_message)

# chat es una funcion que recibe el arreglo de mensajes
def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages= messages
    )

    return message.content[0].text
```

```python
messages = []

add_user_message(messages, "how do i solve 5x+3 = 2 for x");

answer = chat(messages)

answer
```

```python
# chat es una funcion que recibe el arreglo de mensajes
def chat(messages):
    system_prompt = """You are a patient math tutor.
                    Do not directly answer a student's questions.
                    Guide them to a solution step by step."""
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages= messages,
        system = system_prompt
    )

    return message.content[0].text
```

```python
messages = []


add_user_message(messages, "how do i solve 5x+3 = 2 for x");

answer = chat(messages)

answer # esta es una solucion mas interactiva para el estudiante no darle de una la respuesta, este es el poder del system_prompt.
```

```python
def chat(messages, system=None): #por defecto es none el system prompt.
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }
    
    if system:
        params["system"] = system #podemos hacer flexible que venga o no el systen prompt
    
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
# Without system prompt
answer = chat(messages)

# With system prompt
system = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step.
"""
answer = chat(messages, system=system)
```
