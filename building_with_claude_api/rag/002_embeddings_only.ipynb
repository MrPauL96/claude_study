```python
from dotenv import load_dotenv
import voyageai

load_dotenv()
vo = voyageai.Client()
```

```python
chunks = [
    "## Section 1: Overview\nThe system handles real-time events and async jobs.",
    "## Section 2: Security\nRequests use short-lived tokens validated at the gateway.",
    "## Section 3: Retries\nClients retry 429 responses with exponential backoff."
]

def generate_embedding(text, model="voyage-3-large"):
    response = vo.embed([text], model=model, input_type="document")
    return response.embeddings[0]

first_embedding = generate_embedding(chunks[0])
print("Length of embedding:", len(first_embedding))
print("First 6 values:", first_embedding[:6])
```

```python
# Generar embeddings para todos los chunks de una sola vez
batch_response = vo.embed(chunks, model="voyage-3-large", input_type="document")
all_embeddings = batch_response.embeddings

print("Total chunks embedded:", len(all_embeddings))
print("Vector dimension:", len(all_embeddings[0]))
```

```python
import re

def chunk_by_section(document_text):
    pattern = r"\n## " # agrupa los textos por patron de markdown... con pattern.
    return re.split(pattern, document_text.strip())
```

```python
with open("./report.md", "r") as f:
    text = f.read()

chunks = chunk_by_section(text)


generate_embedding(chunks[0])
```
