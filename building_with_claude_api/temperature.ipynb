```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# temperatura en 1 es mucho mas creativa...

def chat(messages, system=None, temperature=1.0): #por defecto es none el system prompt.
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature
    }
    
    if system:
        params["system"] = system #podemos hacer flexible que venga o no el systen prompt
    
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
def add_user_message(messages, text):
    user_message = { "role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(message, text):
    assistant_message = { "role": "assistant", "content": text}
    messages.append(assistant_message)
```

```python
messages = []

add_user_message(
    messages,
    "Generate a one sentence movie idea"
)

# 
answer = chat(messages, temperature = 0.0) 

answer
```
